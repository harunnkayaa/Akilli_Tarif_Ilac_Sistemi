```python
import re
import json
import unicodedata
from difflib import SequenceMatcher

import pandas as pd

# =========================
# 1) DOSYA YOLLARINI GİR
# =========================
NUTRITION_PATH = "/Users/harunkaya/Downloads/Nutritional Values Final1.xlsx"
RECIPES_PATH   = "/Users/harunkaya/Downloads/tarifler.xlsx"

NUTRITION_COL = "canonical_name_tr"
RECIPES_COL   = "Malzemeler_JSON (Stok Takibi İçin)"

FUZZY_THRESHOLD = 0.86

# =========================
# 2) OKUMA
# =========================
def read_any(path: str) -> pd.DataFrame:
    if path.lower().endswith(".csv"):
        return pd.read_csv(path)
    if path.lower().endswith(".xlsx") or path.lower().endswith(".xls"):
        return pd.read_excel(path)
    raise ValueError(f"Desteklenmeyen dosya uzantısı: {path}")

nutrition_df = read_any(NUTRITION_PATH)
recipes_df = read_any(RECIPES_PATH)

# kolonlarda boşluk varsa temizle
nutrition_df.columns = [c.strip() for c in nutrition_df.columns]
recipes_df.columns   = [c.strip() for c in recipes_df.columns]

if NUTRITION_COL not in nutrition_df.columns:
    raise KeyError(f"{NUTRITION_COL} kolonu Nutrition dosyasında yok. Var olanlar: {nutrition_df.columns.tolist()}")
if RECIPES_COL not in recipes_df.columns:
    raise KeyError(f"{RECIPES_COL} kolonu Tarifler dosyasında yok. Var olanlar: {recipes_df.columns.tolist()}")

# =========================
# 3) NORMALIZE
# =========================
NOISE_WORDS = {
    "taze", "donmuş", "dondurulmuş", "konserve", "kavrulmuş", "haşlanmış", "kızarmış",
    "doğranmış", "rendelenmiş", "ezilmiş", "ince", "kalın", "iri", "küçük",
    "kuşbaşı", "küp", "dilim", "kıyılmış", "öğütülmüş", "süzme", "yağsız", "az", "orta",
    "toz", "granül", "paket", "adet",

    "g", "gr", "kg", "ml", "lt", "l", "cc",
    "yemek", "kaşığı", "yemekkaşığı", "tatlı", "çay", "bardak", "bardağı",
    "su", "çorba", "kasesi", "kase", "demet", "tutam",
}

def tr_lower(s: str) -> str:
    s = s.replace("İ", "i").replace("I", "ı")
    return s.lower()

def normalize_ingredient(name: str) -> str:
    if name is None:
        return ""
    s = str(name).strip()
    if not s:
        return ""

    s = re.sub(r"\([^)]*\)", " ", s)      # parantez içi
    s = tr_lower(s)
    s = unicodedata.normalize("NFKC", s)

    s = re.sub(r"[%\d]", " ", s)          # sayı temizle
    s = re.sub(r"[-_/.,;:!?\"'’`]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()

    tokens = [t for t in s.split(" ") if t and t not in NOISE_WORDS]
    s = " ".join(tokens)

    s = s.replace("yogurt", "yoğurt")
    s = s.replace("kasar", "kaşar")
    s = s.replace("zeytinyagi", "zeytinyağı")

    return s.strip()

# =========================
# 4) TARİFLERDEN MALZEME ÇEKME (JSON/Liste/String)
# =========================
def parse_malzemeler_cell(cell):
    """
    Tarifler kolonundaki hücre örneğin şöyle:
    [
      {"Malzeme_Adi":"Pirinç", "Tarif_Olcum":"1 su bardağı", ...},
      {"Malzeme_Adi":"Su", ...}
    ]
    Bu fonksiyon sadece Malzeme_Adi'leri çıkarır.
    """
    if cell is None or (isinstance(cell, float) and pd.isna(cell)) or (isinstance(cell, str) and not cell.strip()):
        return []

    # excel bazen string verir bazen direkt list verir; ikisini de destekle
    if isinstance(cell, list):
        obj = cell
    else:
        s = str(cell).strip()

        # JSON gibi duruyorsa parse et
        if (s.startswith("[") and s.endswith("]")) or (s.startswith("{") and s.endswith("}")):
            try:
                obj = json.loads(s)
            except Exception:
                obj = None
        else:
            obj = None

    # JSON parse edilemediyse, en son çare: düz metin bölme
    if obj is None:
        s = str(cell)
        parts = re.split(r"[\n\r,;•\-]\s*", s)
        return [p.strip() for p in parts if p and p.strip()]

    # Liste ise içinden isimleri çek
    if isinstance(obj, list):
        out = []
        for item in obj:
            if isinstance(item, str):
                out.append(item)
            elif isinstance(item, dict):
                # ✅ EN KRİTİK DÜZELTME: Malzeme_Adi key'i
                for key in ["Malzeme_Adi", "malzeme_adi", "Name", "name", "malzeme", "ingredient", "canonical_name_tr", "item"]:
                    if key in item and item[key]:
                        out.append(str(item[key]))
                        break
            else:
                out.append(str(item))
        return out

    # Dict ise değerleri al
    if isinstance(obj, dict):
        # önce olası isim key'lerini dene
        for key in ["Malzeme_Adi", "malzeme_adi", "Name", "name", "malzeme", "ingredient", "canonical_name_tr", "item"]:
            if key in obj and obj[key]:
                return [str(obj[key])]
        return [str(v) for v in obj.values() if v]

    return []

# Tariflerdeki tüm malzemeler (raw)
recipe_ingredients_raw = []
for cell in recipes_df[RECIPES_COL].tolist():
    recipe_ingredients_raw.extend(parse_malzemeler_cell(cell))

# normalize edip benzersiz
recipe_norm_set = set()
recipe_norm_to_raw = {}
for raw in recipe_ingredients_raw:
    norm = normalize_ingredient(raw)
    if not norm:
        continue
    recipe_norm_set.add(norm)
    recipe_norm_to_raw.setdefault(norm, raw)

# Nutrition tarafı (canonical) normalize
nutrition_df = nutrition_df.dropna(subset=[NUTRITION_COL])
nutrition_raw = nutrition_df[NUTRITION_COL].astype(str).str.strip().tolist()

nutri_norm_set = set()
nutri_norm_to_raw = {}
for raw in nutrition_raw:
    norm = normalize_ingredient(raw)
    if not norm:
        continue
    nutri_norm_set.add(norm)
    nutri_norm_to_raw.setdefault(norm, raw)

# =========================
# 5) EKSİKLERİ BUL + FUZZY
# =========================
def sim(a: str, b: str) -> float:
    return SequenceMatcher(None, a, b).ratio()

missing = []
for r_norm in sorted(recipe_norm_set):
    if r_norm in nutri_norm_set:
        continue

    best_norm = None
    best_score = 0.0
    for n_norm in nutri_norm_set:
        score = sim(r_norm, n_norm)
        if score > best_score:
            best_score = score
            best_norm = n_norm

    status = "fuzzy_matched" if best_score >= FUZZY_THRESHOLD else "missing"

    missing.append({
        "recipe_ingredient_raw_example": recipe_norm_to_raw.get(r_norm, ""),
        "recipe_ingredient_normalized": r_norm,
        "closest_nutrition_match_raw": nutri_norm_to_raw.get(best_norm, "") if best_norm else "",
        "closest_nutrition_match_normalized": best_norm or "",
        "similarity": round(best_score, 3),
        "status": status
    })

missing_df = pd.DataFrame(missing)

true_missing_df = missing_df[missing_df["status"] == "missing"].copy().sort_values(
    ["similarity", "recipe_ingredient_normalized"], ascending=[True, True]
)

# =========================
# 6) ÇIKTI
# =========================
print("Tariflerden çekilen raw malzeme sayısı:", len(recipe_ingredients_raw))
print("Toplam tarif malzemesi (normalize, unique):", len(recipe_norm_set))
print("Toplam nutrition canonical (normalize, unique):", len(nutri_norm_set))
print("Eşleşmeyen (toplam):", len(missing_df))
print("Gerçek eksik (threshold altında):", len(true_missing_df))

missing_df.to_csv("ingredient_match_report_all.csv", index=False, encoding="utf-8-sig")
true_missing_df.to_csv("ingredient_missing_only.csv", index=False, encoding="utf-8-sig")

with pd.ExcelWriter("ingredient_match_report.xlsx") as w:
    missing_df.to_excel(w, sheet_name="all_unmatched", index=False)
    true_missing_df.to_excel(w, sheet_name="missing_only", index=False)

print("\n✅ Raporlar yazıldı:")
print("- ingredient_match_report_all.csv")
print("- ingredient_missing_only.csv")
print("- ingredient_match_report.xlsx")
```
